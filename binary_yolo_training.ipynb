# Binary YOLO Training for Road Damage Detection

YOLOv8m model for binary classification: Road Damage vs No Road Damage

```python
%pip install ultralytics -q

import os
import yaml
import shutil
import random
import torch
from glob import glob
from ultralytics import YOLO

print(f"CUDA available: {torch.cuda.is_available()}")
print(f"GPUs: {torch.cuda.device_count()}")
```

```python
binary_dataset_path = 'yolo_binary_dataset'
binary_augmented_path = 'yolo_binary_augmented_dataset'

if os.path.exists(f"{binary_dataset_path}/data.yaml"):
    print(f"✓ Binary dataset found: {binary_dataset_path}")
else:
    print(f"❌ Binary dataset not found. Run prepare_yolo_dataset.ipynb first")

if os.path.exists(f"{binary_augmented_path}/data.yaml"):
    print(f"✓ Binary augmented dataset found: {binary_augmented_path}")
else:
    print(f"❌ Binary augmented dataset not found")
```

```python
model = YOLO('yolo11n.pt')
```

```python
# dataset_path = binary_augmented_path 
dataset_path = binary_dataset_path 

gpu_ids = [1, 2, 3, 4, 5, 6, 7]
os.environ['CUDA_VISIBLE_DEVICES'] = ','.join(map(str, gpu_ids))
num_gpus = len(gpu_ids)
total_batch_size = 16 * num_gpus

results = model.train(
    data=f'{dataset_path}/data.yaml',
    epochs=150,
    imgsz=640,
    batch=total_batch_size,
    device=gpu_ids,
    project='runs/detect',
    name='binary_road_damage_detection',
    optimizer='Adam',
    cos_lr=True,
    verbose=True
)
```

```python
train_dirs = sorted(glob('runs/detect/binary_road_damage_detection*'), key=os.path.getmtime, reverse=True)
if train_dirs:
    best_model_path = f"{train_dirs[0]}/weights/best.pt"
    eval_model = YOLO(best_model_path)
else:
    eval_model = model

results = eval_model.val(data=f'{dataset_path}/data.yaml', split='test', device=0, verbose=False)

print(f"mAP@0.5: {results.box.map50:.4f}")
print(f"mAP@0.5:0.95: {results.box.map:.4f}")
print(f"Precision: {results.box.mp:.4f}")
print(f"Recall: {results.box.mr:.4f}")
print(f"F1-Score: {2 * (results.box.mp * results.box.mr) / (results.box.mp + results.box.mr):.4f}")
```

```python
train_dirs = sorted(glob('runs/detect/binary_road_damage_detection*'), key=os.path.getmtime, reverse=True)
if train_dirs:
    best_model_path = f"{train_dirs[0]}/weights/best.pt"
    os.makedirs('binary_yolo_checkpoints', exist_ok=True)
    shutil.copy(best_model_path, 'binary_yolo_checkpoints/best_binary_road_damage.pt')
    print("Model saved: binary_yolo_checkpoints/best_binary_road_damage.pt")
```

```python
from pathlib import Path
import matplotlib.pyplot as plt
import cv2

def load_best_binary_model():
    if os.path.exists('binary_yolo_checkpoints/best_binary_road_damage.pt'):
        return YOLO('binary_yolo_checkpoints/best_binary_road_damage.pt')
    train_dirs = sorted(glob('runs/detect/binary_road_damage_detection*'), key=os.path.getmtime, reverse=True)
    if train_dirs:
        return YOLO(f"{train_dirs[0]}/weights/best.pt")
    return YOLO('yolov8m.pt')

def visualize_predictions(model, image_path, conf_threshold=0.5):
    results = model(image_path, device=0, conf=conf_threshold)
    annotated_img = results[0].plot()
    
    plt.figure(figsize=(12, 8))
    plt.imshow(cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB))
    plt.title(f"Detection: {Path(image_path).name}")
    plt.axis('off')
    
    boxes = results[0].boxes
    if boxes is not None and len(boxes) > 0:
        print(f"Detections: {len(boxes)}, Max confidence: {boxes.conf.max():.3f}")
    else:
        print("No detections")
    return results

# Run inference
trained_model = load_best_binary_model()
test_images = glob(f'{dataset_path}/test/images/*.jpg')[:3]

for img_path in test_images:
    visualize_predictions(trained_model, img_path, conf_threshold=0.3)
    plt.show()
```

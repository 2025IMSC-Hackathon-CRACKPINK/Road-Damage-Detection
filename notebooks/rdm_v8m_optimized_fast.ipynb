```python
from ultralytics import YOLO

# 모델 초기화 (YOLOv8m)
model = YOLO('yolov8m.pt')

# 학습 실행
model.train(
    data='/content/drive/MyDrive/USC-Hackathon/project/data.yaml',
    epochs=100,                # Epoch 줄이고 성능 유지 (속도 향상)
    imgsz=640,                 # 이미지 크기 축소 (896 -> 640)
    batch=16,                  # Batch size 증가 (메모리 확인 필요)
    project='/content/drive/MyDrive/USC-Hackathon/project',
    name='rdm_v8m_optimized_fast',
    pretrained=True,           # 사전 학습 모델 사용
    save=True,
    save_period=10,

    # 손실 가중치
    box=5.0,
    cls=2.0,
    dfl=1.5,

    # 증강 전략
    mosaic=1.0,
    mixup=0.2,                 # mixup 활성화
    copy_paste=0.1,            # copy-paste 추가
    hsv_h=0.1,
    hsv_v=0.1,
    hsv_s=0.1,
    auto_augment='randaugment',

    # 학습 안정화
    amp=True,                  # mixed precision
    cache=True,                # 데이터 캐싱
    patience=20,               # 조기 종료
    cos_lr=True                # Cosine Learning Rate 스케줄링
)
```

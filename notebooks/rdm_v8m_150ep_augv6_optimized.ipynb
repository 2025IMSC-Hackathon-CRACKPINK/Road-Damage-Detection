```python
from ultralytics import YOLO

# 모델 초기화 (YOLOv8m)
model = YOLO('yolov8m.pt')

# 학습 시작
model.train(
    data='/content/drive/MyDrive/USC-Hackathon/project/data.yaml',
    epochs=150,
    imgsz=896,
    batch=8,
    project='/content/drive/MyDrive/USC-Hackathon/project',
    name='rdm_v8m_150ep_augv6_optimized',
    pretrained=True,
    save=True,
    save_period=10,

    # 손실 가중치
    box=5.0,
    cls=2.0,
    dfl=1.5,

    # 증강 전략
    augment=True,
    auto_augment='randaugment',
    mosaic=1.0,
    mixup=0.0,
    copy_paste=0.0,
    hsv_h=0.1,
    hsv_v=0.1,
    hsv_s=0.1,

    # 학습 안정화
    amp=True,
    cache=True,
    patience=30
)
```

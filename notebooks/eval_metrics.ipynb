```python
import pandas as pd

# 결과 파일 경로
csv_path = '/content/drive/MyDrive/USC-Hackathon/project/rdm_v8m_150ep_augv6_optimized_v2/results.csv'
df = pd.read_csv(csv_path)

# 🔍 mAP50 기준 best epoch 찾기 (또는 mAP50-95 사용 가능)
best_epoch_idx = df['metrics/mAP50(B)'].idxmax()
best_row = df.iloc[best_epoch_idx]

# precision & recall 추출
precision = best_row['metrics/precision(B)']
recall = best_row['metrics/recall(B)']

# F1 Score 계산
if precision + recall > 0:
    f1_score = 2 * (precision * recall) / (precision + recall)
    print(f"✅ [Best Epoch {int(best_row['epoch'])}] F1 Score: {f1_score:.4f}")
else:
    print("⚠️ Precision + Recall = 0, F1 계산 불가")
```

```python
import os

# 폴더 경로
train_images = '/content/drive/MyDrive/USC-Hackathon/project/images/train'
val_images   = '/content/drive/MyDrive/USC-Hackathon/project/images/val'
test_images  = '/content/drive/MyDrive/USC-Hackathon/images/test'

def list_files(directory):
    return set([os.path.splitext(f)[0] for f in os.listdir(directory) if f.endswith('.jpg')])

train_files = list_files(train_images)
val_files   = list_files(val_images)
test_files  = list_files(test_images)

# 교집합 확인
train_val_overlap  = train_files & val_files
train_test_overlap = train_files & test_files
val_test_overlap   = val_files & test_files

print("=== 데이터 누수 여부 체크 ===")
print(f"Train-Val 중복 수: {len(train_val_overlap)}")
print(f"Train-Test 중복 수: {len(train_test_overlap)}")
print(f"Val-Test 중복 수: {len(val_test_overlap)}")

# 필요 시 중복 리스트 출력
if train_val_overlap:
    print("\nTrain-Val 중복 예시:", list(train_val_overlap)[:10])
if train_test_overlap:
    print("\nTrain-Test 중복 예시:", list(train_test_overlap)[:10])
if val_test_overlap:
    print("\nVal-Test 중복 예시:", list(val_test_overlap)[:10])
```

```python
import os
from glob import glob

# 경로 설정
base_dir = '/content/drive/MyDrive/USC-Hackathon/images'
train_dir = os.path.join(base_dir, 'train')
val_dir   = os.path.join(base_dir, 'val')
test_dir  = os.path.join(base_dir, 'test')

def list_files(directory):
    return set([os.path.splitext(f)[0] for f in os.listdir(directory) if f.endswith('.jpg')])

def count_aug_files(directory):
    return len([f for f in os.listdir(directory) if '_aug' in f and f.endswith('.jpg')])

# 파일 목록 불러오기
train_files = list_files(train_dir)
val_files   = list_files(val_dir)
test_files  = list_files(test_dir)

# 중복 체크
train_val_overlap  = train_files & val_files
train_test_overlap = train_files & test_files
val_test_overlap   = val_files & test_files

print("=== 데이터 중복 여부 ===")
print(f"Train-Val 중복 수 : {len(train_val_overlap)}")
print(f"Train-Test 중복 수: {len(train_test_overlap)}")
print(f"Val-Test 중복 수  : {len(val_test_overlap)}")

if train_val_overlap:
    print(f"  Train-Val 중복 예시: {list(train_val_overlap)[:10]}")
if train_test_overlap:
    print(f"  Train-Test 중복 예시: {list(train_test_overlap)[:10]}")
if val_test_overlap:
    print(f"  Val-Test 중복 예시: {list(val_test_overlap)[:10]}")

# 증강 데이터 개수 확인
train_aug_count = count_aug_files(train_dir)
val_aug_count   = count_aug_files(val_dir)
test_aug_count  = count_aug_files(test_dir)

print("\n=== 증강 데이터 개수 ===")
print(f"Train 폴더 내 증강 이미지: {train_aug_count}개")
print(f"Val   폴더 내 증강 이미지: {val_aug_count}개")
print(f"Test  폴더 내 증강 이미지: {test_aug_count}개")

# 각 폴더 전체 이미지 개수
print("\n=== 각 폴더 이미지 총 개수 ===")
print(f"Train: {len(train_files)}장")
print(f"Val  : {len(val_files)}장")
print(f"Test : {len(test_files)}장")
```

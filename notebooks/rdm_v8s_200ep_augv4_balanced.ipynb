```python
from ultralytics import YOLO

model = YOLO('yolov8s.pt')  # YOLOv8s: 성능 중심 선택

model.train(
    data='/content/drive/MyDrive/USC-Hackathon/project/data.yaml',
    epochs=200,                          # 학습 지속 시간 증가
    imgsz=896,                           # 고해상도 유지
    batch=16,
    project='/content/drive/MyDrive/USC-Hackathon/project',
    name='rdm_v8s_200ep_augv4_balanced',
    pretrained=True,
    save=True,

    # 손실 함수 비중
    box=6.0,
    cls=1.5,
    dfl=1.0,

    # 증강 전략
    augment=True,
    auto_augment='randaugment',
    mosaic=1.0,
    mixup=0.2,
    copy_paste=0.3,
    hsv_h=0.05,
    hsv_v=0.1,

    # 학습 안정화
    cache=True,
    amp=True,
    patience=40
)
```

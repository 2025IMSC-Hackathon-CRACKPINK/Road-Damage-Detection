### 클래스별 개수 확인

```python
import os
from collections import defaultdict

# 기존: 증강 전 레이블 디렉토리
label_dir = '/content/drive/MyDrive/USC-Hackathon/labels/train'

# 수정: 증강된 레이블 디렉토리
# label_dir = '/content/drive/MyDrive/USC-Hackathon/project/labels/train_aug'


class_counts = defaultdict(int)

for file in os.listdir(label_dir):
    if not file.endswith('.txt'):
        continue
    with open(os.path.join(label_dir, file), 'r') as f:
        for line in f:
            class_id = int(line.strip().split()[0])
            class_counts[class_id] += 1

# 클래스명 매핑
class_names = {
    0: 'Pothole',
    1: 'Alligator Crack',
    2: 'Transverse Crack',
    3: 'Longitudinal Crack',
}

print("Train Set 클래스별 바운딩 박스 수:")
for cls_id in sorted(class_counts):
    print(f"  {class_names[cls_id]:20}: {class_counts[cls_id]}개")
```
